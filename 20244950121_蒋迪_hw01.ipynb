```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

#测试三组数据
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))
```

### 负数参数思考
当输入负数，Python取模运算会保留符号，函数仍然返回正数结果。
但数学上最大公约数GCD定义针对非负整数，输入负数的行为没有数学意义，本函数没有做输入合法性校验。
测试示例 gcd(-12, 8)

```python
import math

# 1.极限近似 e
def euler_limit(n):
    return (1 + 1 / n) ** n

#2.级数前n项求和
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1
    for k in range(n):
        if k == 0:
            fact = 1
        else:
            fact *= k
        total += 1.0 / fact
    return total

#3.按误差epsilon逼近e，while循环
def euler_approx(epsilon):
    total = 0.0
    k = 0
    while abs(total - math.exp(1)) > epsilon:
        fact = 1
        for i in range(1, k+1):
            fact *= i
        total += 1.0 / fact
        k += 1
    return total

#4打印表格
def print_euler_sum_table(n):
    for k in range(1, n+1):
        val = euler_infinite_sum(k)
        print(f"k={k:2d}  sum={val:.6f}")

#简单测试
print(euler_infinite_sum(1))
print(euler_infinite_sum(4))
print_euler_sum_table(5)
```

```python
def is_power_of_2(n):
    while n > 1:
        if n % 2 != 0:
            return False
        n = n // 2
    return True

def is_power(b, n):
    while n > 1:
        if n % b != 0:
            return False
        n = n // b
    return True

#测试
print(is_power_of_2(8))
print(is_power_of_2(7))
print(is_power(3,81))
print(is_power(3,82))
```

```python
# 1分数等级函数
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须处于0~100范围")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

# 2 for、while求1~100总和
sum_for = 0
for num in range(1,101):
    sum_for += num
print("for循环求和1‑100：", sum_for)

sum_while = 0
n = 1
while n <=100:
    sum_while += n
    n +=1
print("while循环求和1‑100：", sum_while)

# 3迭代阶乘、递归阶乘
def fact_iter(n):
    res = 1
    for i in range(1,n+1):
        res *= i
    return res

def fact_recur(n):
    if n == 0:
        return 1
    return n * fact_recur(n-1)

for test_n in [0,1,5,10]:
    print(f"n={test_n} 迭代:{fact_iter(test_n)} 递归:{fact_recur(test_n)}")

# 4 summary 返回元组，元组解包
def summary(values):
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values)/len(values)
    return (min_v, max_v, avg_v)

ret = summary([2,4,6,8])
mi, ma, avg = ret
print("summary结果：",mi,ma,avg)

# 5可变参数 sum_all
def sum_all(*args):
    return sum(args)

print("无参数：", sum_all())
print("单参数：", sum_all(9))
print("多参数：", sum_all(1,2,3,4,5))
```

```python
import os
import pickle

# 4.1 查看工作目录，创建文件夹
cwd = os.getcwd()
print("当前工作目录：", cwd)

data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据文件夹路径：", data_dir)

# 4.2 写入 students.txt
txt_content = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

txt_path = os.path.join(data_dir, "students.txt")
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(txt_content)

# 读取文件，跳过表头
with open(txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

print("原始全部行：")
for line in lines:
    print(repr(line))
```

```python
# 4.3 parse_student 解析一行，非法抛 ValueError
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段")
    name, score_str = parts
    name = name.strip()
    score_str = score_str.strip()
    if len(name) == 0:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩不是整数：{score_str}")
    if not (0 <= score <= 100):
        raise ValueError(f"成绩超出0‑100范围：{score}")
    return {"name": name, "score": score}


# 简单自测
# print(parse_student("Alice,85"))   #正常
# print(parse_student("Grace,abc")) #会抛异常
```

```python
# 4.4 逐行处理文件，捕获异常
valid_records = []
error_info = []

# lines来自上一个cell读取的全部行，跳过第0行表头
for row_idx, raw_line in enumerate(lines[1:], start=2):
    try:
        rec = parse_student(raw_line)
        valid_records.append(rec)
    except ValueError as e:
        error_info.append({"line": row_idx, "reason": str(e)})

print("✅有效记录：")
for r in valid_records:
    print(r)

print("\n❌错误记录：")
for err in error_info:
    print(err)

# 统计
valid_num = len(valid_records)
invalid_num = len(error_info)
if valid_num > 0:
    avg_score = sum(r["score"] for r in valid_records) / valid_num
    pass_cnt = sum(1 for r in valid_records if r["score"] >= 60)
    pass_rate = pass_cnt / valid_num
else:
    avg_score = 0
    pass_rate = 0
```

```python
# 4.5 report.txt 写入，演示 w覆盖 / a追加
report_path = os.path.join(data_dir, "report.txt")

# w模式：覆盖原有文件，旧内容全部删除
with open(report_path, "w", encoding="utf-8") as f:
    f.write(f"有效记录数：{valid_num}\n")
    f.write(f"无效记录数：{invalid_num}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.1%}\n")

# a模式：追加，在文件末尾新增文字，不会删除旧内容
with open(report_path, "a", encoding="utf-8") as f:
    f.write("---本条是a模式追加的文本---\n")

# 读取打印报告看效果
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())
```

```python
import copy

# 5.1 Student类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


# 5.2 根据有效记录创建Student对象
stu_obj_list = []
for rec in valid_records:
    s = Student(rec["name"], rec["score"])
    stu_obj_list.append(s)
    print(s, "及格状态:", s.is_passed())


# 5.3 GraduateStudent子类继承
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},research='{self.research_topic}')"

# 测试子类
g_stu = GraduateStudent("Tom", 88, "大模型学习")
print(g_stu)
print("是否及格", g_stu.is_passed())


# 5.4 直接赋值 / copy.copy浅拷贝 / copy.deepcopy深拷贝
# 原始嵌套对象，每次测试都从原始对象开始
original = {"name": "Alice", "scores": [85, 90]}

# 1 直接赋值
obj1 = original
# 2 浅拷贝
obj2 = copy.copy(original)
# 3 深拷贝
obj3 = copy.deepcopy(original)

# 修改内部嵌套列表
obj1["scores"][0] = 999

print("original:", original)
print("直接赋值obj1:", obj1)
print("浅拷贝obj2:", obj2)
print("深拷贝obj3:", obj3)

'''
现象解释：
直接赋值：只是同一个对象别名，全部一起变
copy.copy浅拷贝：字典是新对象，但内部嵌套列表仍然共享，会跟着变
copy.deepcopy深拷贝：所有嵌套对象全部复制，修改互不影响
'''
```
